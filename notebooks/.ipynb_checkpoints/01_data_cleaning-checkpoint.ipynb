```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel("../data/Healthcare_dataset.xlsx")

df.head()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.shape
```

```python
df["Date of Admission"] = pd.to_datetime(df["Date of Admission"])
df["Discharge Date"] = pd.to_datetime(df["Discharge Date"])
```

```python
df.info()
```

```python
df["length_of_stay"] = (
    df["Discharge Date"] - df["Date of Admission"]
).dt.days
```

```python
df[["Date of Admission","Discharge Date","length_of_stay"]].head()
```

```python
df.to_csv("../data/cleaned_healthcare_data.csv", index=False)
```

```python
df = df.drop(columns=[
"Patient ID",
"Name",
"Doctor",
"Hospital",
"Room Number"
])
```

```python
df.columns
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
encoder = LabelEncoder()
```

```python
categorical_cols = [
"Gender",
"Blood Type",
"Medical Condition",
"Insurance Provider",
"Admission Type",
"Test Results",
"Medication",
"Severity",
"Dosage",
"Treatment Outcome"
]

for col in categorical_cols:
    df[col] = encoder.fit_transform(df[col])
```

```python
df.head()
```

```python
df = df.drop(columns=[
"Date of Admission",
"Discharge Date"
])
```

```python
df.info()
```

```python
df.to_csv("../data/ml_ready_data.csv", index=False)
```
